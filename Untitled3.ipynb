```python
# -----------------------------------------------------------------
# 1. Instalar as bibliotecas necessárias
# -----------------------------------------------------------------
# O 'ultralytics' é o YOLOv8, e 'roboflow' é para baixar os dados
!pip install ultralytics roboflow

# -----------------------------------------------------------------
# 2. Baixar o seu Dataset do Roboflow
# -----------------------------------------------------------------
# !!! SUBSTITUA AS LINHAS ABAIXO !!!
# Cole aqui o trecho de código que você copiou do Roboflow (Passo 1)

from roboflow import Roboflow

rf = Roboflow(api_key="iToLkCfKPf3yiKTCYDzy")
project = rf.workspace("lab2-3ztow").project("hockey-4dgoj")
version = project.version(1)
dataset = version.download("yolov8")


# -----------------------------------------------------------------
# 3. Iniciar o Treinamento do YOLOv8
# -----------------------------------------------------------------
# Importamos a biblioteca YOLO
from ultralytics import YOLO

# Carregamos um modelo base (yolov8n.pt).
# Ele já sabe identificar formas básicas e nós o especializamos em hockey.
model = YOLO('yolov8n.pt')

# Inicia o treinamento!
# 'data': aponta para o arquivo .yaml que o Roboflow baixou (geralmente data.yaml)
# 'epochs': quantas vezes o modelo vai "ler" o dataset. 50 é um bom começo.
# 'imgsz': o tamanho das imagens durante o treino.
results = model.train(data=f'{dataset.location}/data.yaml', epochs=50, imgsz=640)

print("Treinamento concluído!")
print(f"Seus pesos estão em: {results.save_dir}")
```
